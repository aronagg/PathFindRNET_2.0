# 03 Homography transform
Apply a sample homography matrix to image-plane trajectory points.


```python
import json
import numpy as np
import pandas as pd
df = pd.read_csv('../sample_data/sample_trajectories.csv')
with open('../sample_data/sample_homography.json') as f:
    H = np.array(json.load(f)['matrix'], dtype=float)
H
```

```python
def apply_homography(x, y, H):
    p = np.array([x, y, 1.0])
    q = H @ p
    q = q / q[2]
    return q[0], q[1]

xy = df.apply(lambda r: apply_homography(r['x_image'], r['y_image'], H), axis=1)
df[['x_homography', 'y_homography']] = pd.DataFrame(xy.tolist(), index=df.index)
df.head()
```
