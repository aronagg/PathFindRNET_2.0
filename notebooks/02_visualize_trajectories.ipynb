# 02 Visualize trajectories
Plot sample image-plane trajectories. Replace the sample CSV with real exported trajectories after processing.


```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('../sample_data/sample_trajectories.csv')
df.head()
```

```python
fig, ax = plt.subplots(figsize=(7, 5))
for track_id, g in df.groupby('track_id'):
    ax.plot(g['x_image'], g['y_image'], marker='o', label=f'track {track_id}')
ax.invert_yaxis()
ax.set_xlabel('x image [px]')
ax.set_ylabel('y image [px]')
ax.set_title('Sample image-plane trajectories')
ax.legend()
plt.show()
```
