# 05 Filtering
Demonstrates trajectory filtering by length, class, parking/low-speed behaviour and application-specific ROI/movement rules.


```python
import pandas as pd
import numpy as np
df = pd.read_csv('../sample_data/sample_trajectories_filtering.csv')
df.head()
```

```python
features = []
for tid, g in df.sort_values('frame_id').groupby('track_id'):
    first, last = g.iloc[0], g.iloc[-1]
    features.append({
        'track_id': tid,
        'point_count': len(g),
        'class_name': g['class_name'].mode().iat[0],
        'mean_speed': g['speed'].mean(),
        'displacement_px': np.hypot(last['x_image']-first['x_image'], last['y_image']-first['y_image']),
        'roi_ratio': g['inside_application_roi'].mean(),
        'start_y': first['y_image'],
        'end_y': last['y_image']
    })
features = pd.DataFrame(features)
features
```

```python
filtered = features[
    (features['point_count'] >= 4) &
    (features['class_name'] == 'car') &
    (features['displacement_px'] >= 25) &
    (features['mean_speed'] >= 0.5) &
    (features['roi_ratio'] >= 0.75) &
    (features['end_y'] < features['start_y'])
]
filtered
```
